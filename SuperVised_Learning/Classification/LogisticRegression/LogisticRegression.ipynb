# 📊 Logistic Regression (Binary Classification)

## Introduction

Machine Learning problems are broadly divided into two categories:

1. **Regression** – Predicts continuous numerical values.
2. **Classification** – Predicts categorical labels or classes.

Logistic Regression is one of the most popular **Supervised Machine Learning Classification algorithms**.

Despite its name containing the word **Regression**, Logistic Regression is actually used for **Classification Problems** because it predicts the **probability** of an observation belonging to a particular class.

For example:

- Email → Spam or Not Spam
- Student → Pass or Fail
- Customer → Buy or Not Buy
- Patient → Disease or No Disease

Unlike Linear Regression, which predicts continuous values, Logistic Regression predicts values between **0 and 1**, representing probabilities.

The predicted probability is then converted into a class label using a threshold (usually **0.5**).

---

## Why Logistic Regression?

Suppose we want to predict whether a student will pass an exam.

Linear Regression may predict values like:

- 1.25
- -0.42
- 2.10

These predictions are not valid probabilities because probabilities must always lie between **0 and 1**.

Logistic Regression solves this problem by using the **Sigmoid Function**, which converts any real number into a probability between **0 and 1**.

---

## Applications

Logistic Regression is widely used in:

- Spam Email Detection
- Disease Prediction
- Loan Approval
- Credit Card Fraud Detection
- Customer Churn Prediction
- Sentiment Analysis
- Employee Attrition Prediction
- Marketing Campaign Prediction

# Step 1: Import Required Libraries

Before building any Machine Learning model, we need to import the required Python libraries.

Libraries used in this notebook:

- matplotlib → Data visualization

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve
)
```

# Step 2: Load the Dataset

We load the Social Network Ads dataset using pandas.

The dataset contains customer information such as:

- Gender
- Age
- Estimated Salary

Our objective is to predict whether the customer purchases the product.

```python
df = pd.read_csv("Social_Network_Ads.csv")

df.head()
```

# Step 3: Explore the Dataset

Before training any Machine Learning model, it is important to understand the dataset.

We use:

- head() → Displays first five rows
- shape → Number of rows and columns
- info() → Data types and missing values
- describe() → Statistical summary

This helps us identify:

- Missing values
- Data types
- Feature distribution
- Dataset size

```python
print(df.head())

print()

print(df.shape)

print()

df.info()

print()

df.describe()
```

# Step 4: Checking Missing Values

Machine Learning models cannot work with missing values.

We verify whether the dataset contains any missing data.

```python
df.isnull().sum()
```

# Step 5: Understanding the Features

Our dataset contains five columns.

| Feature | Description |
|----------|-------------|
| User ID | Unique customer ID |
| Gender | Male or Female |
| Age | Customer Age |
| EstimatedSalary | Annual Salary |
| Purchased | Target Variable |

Notice that **User ID** is simply an identifier.

It does not help the model learn any purchasing pattern.

Therefore, we remove it.

```python
df = df.drop("User ID", axis=1)

df.head()
```

# Step 6: Encoding Categorical Data

Machine Learning algorithms work only with numbers.

The Gender column contains text values.

Male

Female

We convert them into numbers using Label Encoding.

Example

Male → 1

Female → 0

```python
encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

df.head()
```

# Step 7: Defining Features and Target

Machine Learning datasets are divided into:

X → Independent Variables (Features)

y → Dependent Variable (Target)

Features:

- Gender
- Age
- EstimatedSalary

Target:

Purchased

```python


X = df.drop("Purchased", axis=1)

y = df["Purchased"]

```

# Step 8: Splitting the Dataset

We split the dataset into:

Training Data → 80%

Testing Data → 20%

The model learns from the training data and is evaluated on the testing data.

This helps us measure how well the model performs on unseen data.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

# Step 9: Feature Scaling

Logistic Regression performs better when numerical features are on a similar scale.

Example

Age = 20

Salary = 90000

Since Salary has much larger values, it may dominate the learning process.

StandardScaler transforms both features to have similar scales.

This improves model convergence and performance.

```python

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

# Step 10: Creating the Logistic Regression Model

Now we create the Logistic Regression model.

Initially, the model does not know any relationship between the features and the target.

It will learn this relationship during training.

```python
model = LogisticRegression(random_state=42)
```

# Step 11: Training the Model

Training means teaching the model using historical data.

During training, Logistic Regression:

1. Initializes random weights
2. Calculates probabilities
3. Computes Log Loss
4. Updates weights
5. Repeats until convergence

The learned weights are stored inside the model.

```python
model.fit(X_train, y_train)
```

# Step 12: Making Predictions

After training, we use the testing dataset to predict whether customers purchase the product.

The model returns either:

0 → Not Purchased

1 → Purchased

```python
y_pred = model.predict(X_test)

y_pred[:10]
```

# Step 13: Predicting Probabilities

Instead of directly predicting classes, Logistic Regression first predicts probabilities.

For example:

[0.92 0.08]

means

92% probability of class 0

8% probability of class 1

The class with the higher probability becomes the prediction.

```python
probabilities = model.predict_proba(X_test)

print(probabilities[:5])
```

# Evaluation Metrics for Classification

Unlike Regression problems, Classification models cannot be evaluated using metrics such as Mean Squared Error (MSE) alone.

Instead, we use specialized Classification Metrics.

The most commonly used metrics are:

- Accuracy
- Confusion Matrix
- Precision
- Recall
- F1 Score
- ROC Curve
- AUC Score

Each metric provides different information about model performance.

Choosing the correct evaluation metric depends on the problem being solved.

# Accuracy

Accuracy measures the overall correctness of the model.

It is the ratio of correctly predicted observations to the total observations.

Formula

Accuracy = (TP + TN) / (TP + TN + FP + FN)

Where

TP = True Positive

TN = True Negative

FP = False Positive

FN = False Negative

Example

Suppose:

Correct Predictions = 95

Total Predictions = 100

Accuracy = 95 / 100

Accuracy = 95%

Higher Accuracy generally indicates better model performance.

However, Accuracy may be misleading for imbalanced datasets.

# Confusion Matrix

A Confusion Matrix summarizes the performance of a Classification model.

It consists of four values.

|                | Predicted Positive | Predicted Negative |
|----------------|-------------------|-------------------|
| Actual Positive | True Positive (TP) | False Negative (FN) |
| Actual Negative | False Positive (FP) | True Negative (TN) |

Meaning of each term

True Positive (TP)

The model correctly predicts the positive class.

Example:

Patient has cancer

Model predicts cancer

True Negative (TN)

The model correctly predicts the negative class.

Example:

Patient is healthy

Model predicts healthy

False Positive (FP)

The model predicts positive but the actual class is negative.

Example:

Healthy patient predicted as cancer.

False Negative (FN)

The model predicts negative but the actual class is positive.

Example:

Cancer patient predicted as healthy.

False Negatives are usually more dangerous in medical applications.

# Precision

Precision answers the question:

"Out of all positive predictions, how many were actually positive?"

Formula

Precision = TP / (TP + FP)

High Precision means the model makes very few False Positive predictions.

Example Applications

- Spam Detection
- Email Filtering
- Search Engines

# Recall (Sensitivity)

Recall answers the question:

"Out of all actual positive cases, how many did the model correctly identify?"

Formula

Recall = TP / (TP + FN)

High Recall means very few False Negatives.

Recall is extremely important in:

- Disease Prediction
- Fraud Detection
- Security Systems

# F1 Score

F1 Score is the harmonic mean of Precision and Recall.

Formula

F1 = 2 × (Precision × Recall) / (Precision + Recall)

F1 Score becomes useful when:

- Dataset is imbalanced
- Both Precision and Recall are equally important

Range

0 → Worst

1 → Best

# ROC Curve and AUC Score

ROC stands for

Receiver Operating Characteristic Curve.

It compares

True Positive Rate (Recall)

against

False Positive Rate

at different classification thresholds.

The Area Under the ROC Curve (AUC) summarizes the model's ability to distinguish between positive and negative classes.

Interpretation

AUC = 1.0 → Perfect Classifier

AUC = 0.90 → Excellent

AUC = 0.80 → Good

AUC = 0.50 → Random Guessing

Higher AUC indicates a better classification model.

# Step 14: Model Evaluation

We evaluate the model using:

- Accuracy
- Confusion Matrix
- Classification Report
- ROC-AUC Score

These metrics tell us how well the model performs on unseen data.

```python
print("Accuracy :", accuracy_score(y_test, y_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))

print("\nROC-AUC :", roc_auc_score(y_test, model.predict_proba(X_test)[:,1]))
```
